```python
#functions

##function to buld lb

def build_margin_lookup(runners):
    # Create a dictionary {(raceId, ranking): margin}
    return {
        (row.raceId, row.ranking): row.margin
        for row in runners.itertuples(index=False)
    }

def margin_to_length_fast(margin, ranking, raceId, margin_lookup):
    if ranking == 1:
        # Get margin of horse ranked 2
        margin_rank2 = margin_lookup.get((raceId, 2), None)
        if margin_rank2 is None:
            return 0
        return -margin_to_length_fast(margin_rank2, 2, raceId, margin_lookup)

    if pd.isnull(margin):
        return 0

    if isinstance(margin, str):
        margin = margin.replace('(', '').replace(')', '')
        margin_map = {
            'DH': 0, 'NEZ': 0.05, 'CTT': 0.1,
            'TETE': 0.2, 'CTE': 0.3, 'ENC': 0.4,
            'LOIN': 99
        }

        if margin in margin_map:
            return margin_map[margin]

        # Handle mixed fractions and whole numbers
        parts = margin.split()
        length = 0.0
        for part in parts:
            if '/' in part:
                num, denom = part.split('/')
                length += float(num) / float(denom)
            else:
                length += float(part)
        return length

    return float(margin)

runners['draw_unique'] = runners[['draw', 'name_meeting', 'distance', 'direction', 'rail', 'surface']]\
    .apply(lambda row: ' - '.join([str(x) for x in row if pd.notnull(x)]), axis=1)


def rank_horses_by_mention_order(text, horse_names):
    lower_text = text.lower()
    mention_positions = []

    for horse in horse_names:
        pattern = r'\b' + re.escape(horse.lower()) + r'\b'
        match = re.search(pattern, lower_text)
        if match:
            mention_positions.append((match.start(), horse))

    # Sort horses by first occurrence in the text
    mention_positions.sort()
    return {horse: rank + 1 for rank, (_, horse) in enumerate(mention_positions)}

def process_webtips_simple(webTips, runners):
    all_records = []

    for row in tqdm(webTips.itertuples(), total=len(webTips), desc="Processing tips"):
        race_id = row.raceId
        text = row.text
        horse_rows = runners[runners['raceId'] == race_id]
        horse_names = list(horse_rows['horseName'].dropna())

        ranked_horses = rank_horses_by_mention_order(text, horse_names)

        for horse, rank in ranked_horses.items():
            all_records.append({
                "race_id": race_id,
                "runner": horse,
                "webTipRank": rank
            })

    return pd.DataFrame(all_records)


def get_distance_group(m):
    if pd.isna(m):
        return None
    m = int(m)
    if m <= 1000:
        return '0-1000'
    elif m > 3600:
        return '>3600'
    else:
        lower = ((m - 1) // 200) * 200 + 1
        upper = lower + 199
        return f'{lower}-{upper}'

def calculate_prize(race_total, position):
    if position == 1:
        return 0.50 * race_total
    elif position == 2:
        return 0.20 * race_total
    elif position == 3:
        return 0.15 * race_total
    elif position == 4:
        return 0.10 * race_total
    elif position == 5:
        return 0.05 * race_total
    else:
        return 0
```

```python
import textwrap
import pandas as pd

def print_race_details_and_tips(runners_tdy, webTips_tdy, race):
    """
    Print race details and wrapped tips for the given race index.

    Parameters:
    - runners_tdy: DataFrame with runner and race data.
    - webTips_tdy: DataFrame with tips data.
    - race: integer index into runners_tdy.raceId.unique().
    """
    # Get race ID and data
    race_id = runners_tdy.raceId.unique()[race]
    race_data = runners_tdy[runners_tdy.raceId == race_id]

    # helper to convert unique values to clean string
    def clean_join(values):
        return ', '.join(
            str(v) for v in values if pd.notnull(v)
        ) or "N/A"

    meeting_name = clean_join(race_data.meetingName.unique())
    race_type = clean_join(race_data.raceType.unique())
    total_prize = clean_join(race_data.raceTotalPrize.unique())
    distance = clean_join(race_data.distance.unique())
    going = clean_join(race_data.going.unique())

    # Print race details
    print("=" * 80)
    print(f"Meeting:      {meeting_name}")
    print(f"Race Type:    {race_type}")
    print(f"Total Prize:  {total_prize}")
    print(f"Distance:     {distance}")
    print(f"Going:        {going}")
    print("=" * 80)
    print()

    # Get tips
    texts = webTips_tdy[webTips_tdy['raceId'] == race_id].text

    if not texts.empty:
        print("Tips:\n")
        for i, tip in enumerate(texts, 1):
            print(f"{i}.")
            print(textwrap.fill(str(tip), width=150))
            print("-" * 150)
    else:
        print("No tips available for this race.")
```

```python
kg_per_length = {
    "VERY FAST": {"0-1000": 1.429, "1001-1200": 1.334, "1201-1400": 1.191,
                  "1401-1600": 1.072, "1601-1800": 0.952, "1801-2000": 0.905,
                  "2001-2200": 0.810, "2201-2400": 0.762, "2401-2600": 0.714,
                  "2601-2800": 0.667, "2801-3000": 0.619, "3001-3200": 0.571,
                  "3201-3400": 0.524, ">3600": 0.477},
    "FAST": {"0-1000": 1.361, "1001-1200": 1.270, "1201-1400": 1.134,
             "1401-1600": 1.021, "1601-1800": 0.907, "1801-2000": 0.862,
             "2001-2200": 0.771, "2201-2400": 0.726, "2401-2600": 0.680,
             "2601-2800": 0.635, "2801-3000": 0.590, "3001-3200": 0.544,
             "3201-3400": 0.499, ">3600": 0.454},
    "PSF": {"0-1000": 1.388, "1001-1200": 1.295, "1201-1400": 1.156,
            "1401-1600": 1.041, "1601-1800": 0.925, "1801-2000": 0.880,
            "2001-2200": 0.788, "2201-2400": 0.741, "2401-2600": 0.693,
            "2601-2800": 0.646, "2801-3000": 0.598, "3001-3200": 0.551,
            "3201-3400": 0.504, ">3600": 0.463},
    "SLOW": {"0-1000": 1.225, "1001-1200": 1.143, "1201-1400": 1.021,
             "1401-1600": 0.919, "1601-1800": 0.816, "1801-2000": 0.776,
             "2001-2200": 0.694, "2201-2400": 0.653, "2401-2600": 0.612,
             "2601-2800": 0.571, "2801-3000": 0.531, "3001-3200": 0.490,
             "3201-3400": 0.449, ">3600": 0.409},
    "VERY SLOW": {"0-1000": 1.089, "1001-1200": 1.016, "1201-1400": 0.907,
                  "1401-1600": 0.817, "1601-1800": 0.726, "1801-2000": 0.690,
                  "2001-2200": 0.617, "2201-2400": 0.581, "2401-2600": 0.544,
                  "2601-2800": 0.508, "2801-3000": 0.472, "3001-3200": 0.435,
                  "3201-3400": 0.399, ">3600": 0.363},
    None: "use 'FAST' if missing"
}



def lengths_to_rating(l: float) -> float:
    """
    Map a (non‑negative) length difference to a smooth, continuous 1–5 rating
    using a logarithmic‑like curve.

    • At l = 0 → rating = 1.0
    • At l = 6 → rating = 5.0
    • For l ≥ 6 → stays at 5.0
    • Between → smooth increasing curve (log‑like shape)
    """
    if l < 0:
        raise ValueError("Length must be non‑negative")
    if l >= 6.0:
        return 5.0

    # normalize l to [0,1]
    x = l / 6.0
    # log‑like mapping from [0,1] → [0,1]
    y_norm = np.log1p(9 * x) / np.log(10)
    # scale to [1,5]
    return 1.0 + 4.0 * y_norm

def signed_bucket(l_signed):
    """Turn a signed length gap into a signed 1–4 bucket with correct sign."""
    magnitude = lengths_to_rating(abs(l_signed))
    return magnitude if l_signed >= 0 else -magnitude


def update_ratings_with_history(df, k_factor=1.0, alpha=0.07):
    ratings = {}  # horseId → rating
    rating_history = []  # list of dicts to build DataFrame at the end

    df = df.sort_values(by='date')

    # Step 1: Compute prize quantiles per raceId group
    df['prize_quantile'] = df.groupby('raceId')['totalPrize_y'].rank(pct=True)

    race_ids = df['raceId'].unique()
    for race_id in tqdm(race_ids, desc="Processing races"):
        race = df[df['raceId'] == race_id].copy()

        for _, row in race.iterrows():
            h = row['horseId']
            if h not in ratings:
                if 'handicapRatingKg' in row and not pd.isnull(row['handicapRatingKg']):
                    ratings[h] = 30.0 + (row['handicapRatingKg'] - 30.0)
                else:
                    ratings[h] = 30.0

        horses = {
            row['horseId']: {
                'rating': ratings[row['horseId']],
                'lengths_back': row['cumulative_lengths_back'],
                'weight': row['weightKg']
            }
            for _, row in race.iterrows()
        }

        surprises = {h: 0.0 for h in horses}
        n_duels = {h: 0 for h in horses}

        horse_ids = list(horses.keys())
        for i in range(len(horse_ids)):
            for j in range(i + 1, len(horse_ids)):
                h1, h2 = horse_ids[i], horse_ids[j]
                d1, d2 = horses[h1], horses[h2]

                perf_diff = (d1['rating'] - 0.625 * d1['weight']) - (d2['rating'] - 0.625 * d2['weight'])
                bucket_exp = signed_bucket(perf_diff)

                # Use consistent going + distance
                going = row.get('going_category', 'FAST') or 'FAST'
                distance = row.get('distance_group')
                multiplier = kg_per_length.get(going.upper(), kg_per_length['FAST']).get(distance, 1.0)

                length_gap = d2['lengths_back'] - d1['lengths_back']
                actual_diff = length_gap * multiplier

                bucket_act = signed_bucket(actual_diff)

                surprise = bucket_act - bucket_exp
                surprises[h1] += surprise
                surprises[h2] -= surprise
                n_duels[h1] += 1
                n_duels[h2] += 1

        for h in horses:
            avg_surprise = surprises[h] / max(1, n_duels[h])
            horse_row = race[race['horseId'] == h].iloc[0]

            prize_quantile = horse_row.get('prize_quantile', 0.5)
            horse_run = horse_row.get('horse_run', 1)

            # Compute bonus separately (additive)
            prize_bonus = (prize_quantile - 0.5) * 5 * np.exp(-alpha * horse_run)

            # Final rating update
            ratings[h] += k_factor * (avg_surprise + prize_bonus)

            rating_history.append({
                'raceId': race_id,
                'horseId': h,
                'rating_after_race': ratings[h]
            })


    history_df = pd.DataFrame(rating_history)
    df_with_ratings = df.merge(history_df, on=['raceId', 'horseId'], how='left')

    return df_with_ratings
```

```python
# within-race standardization on pos_perc > 0.33
def standardize_within_race(group):
    eligible = group[group['pos_perc'] > 0.33]
    mean = eligible['residual'].mean()
    std = eligible['residual'].std(ddof=0) if eligible['residual'].std(ddof=0) != 0 else 1
    # standardize everyone in the group based on eligible horses’ mean/std
    group['race_adjusted_residual'] = (group['residual'] - mean) / std
    return group
```

```python
def _build_summary(
    df: pd.DataFrame,
    group_by: str
) -> pd.DataFrame:
    summary = (
        df.groupby(group_by)
          .agg(
              runs=("raceId", "count"),
              wins=("win", "sum"),
              places=("place", "sum"),
              sum_odds_win=("odds", "sum"),
              sum_odds_place=("odds_place", "sum"),
              unique_runners=("horseId", pd.Series.nunique),
              avg_pos_perc=("pos_perc", "mean"),
              avg_prizemoney=("prizemoney", "mean"),
              avg_race_adjusted_residual=("race_adjusted_residual", "mean"),
              std_race_adjusted_residual=("race_adjusted_residual", "std"),   # NEW
          )
          .reset_index()
    )


    win_grp = (
        df[df["win"] == 1]
          .groupby(group_by)["horseId"]
          .nunique()
          .reset_index(name="unique_winners")
    )
    summary = (
        summary
          .merge(win_grp, on=group_by, how="left")
          .fillna({"unique_winners": 0})
    )

    horse_means = (
        df.groupby([group_by, "horseId"])["handicapRatingKg"]
          .mean()
          .reset_index()
    )
    avg_hcap = (
        horse_means
          .groupby(group_by)["handicapRatingKg"]
          .mean()
          .reset_index(name="avg_of_avg_handicap")
    )
    summary = summary.merge(avg_hcap, on=group_by, how="left")

    summary["win_pct"]   = (100 * summary["wins"]   / summary["runs"]).round(1)
    summary["place_pct"] = (100 * summary["places"] / summary["runs"]).round(1)

    mask = (
        df["liveOdd"].notnull()
        & df["odds_sum"].ge(1.1)
        & df["odds_sum"].le(1.4)
    )
    ae_df = df[mask]
    ae_grp = (
        ae_df.groupby(group_by)
             .agg(
                 wins_ae=("win", "sum"),
                 sum_odds_win_ae=("odds", "sum"),
                 places_ae=("place", "sum"),
                 sum_odds_place_ae=("odds_place", "sum"),
             )
             .reset_index()
    )
    summary = summary.merge(ae_grp, on=group_by, how="left").fillna({
        "wins_ae": 0,
        "sum_odds_win_ae": 0,
        "places_ae": 0,
        "sum_odds_place_ae": 0,
    })
    summary["ae_win"]   = (summary["wins_ae"]   / summary["sum_odds_win_ae"]).replace([np.inf, np.nan], 0).round(3)
    summary["ae_place"] = (summary["places_ae"] / summary["sum_odds_place_ae"]).replace([np.inf, np.nan], 0).round(3)
    summary.drop(columns=["wins_ae", "sum_odds_win_ae", "places_ae", "sum_odds_place_ae"], inplace=True)

    summary["uniq_win_ratio"] = (summary["unique_winners"] / summary["unique_runners"]).round(3)

    summary[['sum_odds_win', 'sum_odds_place']] = (
        summary[['sum_odds_win', 'sum_odds_place']]
          .replace([np.inf, -np.inf], np.nan)
          .fillna(0)
    )
    summary = summary.astype({
        'runs': int,
        'wins': int,
        'places': int,
        'sum_odds_win': int,
        'sum_odds_place': int,
    })
    summary = summary.round({
        'unique_runners': 1,
        'unique_winners': 1,
        'uniq_win_ratio': 3,
        'avg_pos_perc': 3,
        'avg_prizemoney': 1,
        'avg_of_avg_handicap': 1,
        'avg_race_adjusted_residual': 3,  # NEW
    })

    return summary


def _add_ops(
    summary: pd.DataFrame,
    df_raw: pd.DataFrame,
    group_by: str,
    min_runs: int = 100
) -> pd.DataFrame:
    if df_raw.empty:
        for col in [
            "avg_pos_perc_plus", "avg_prizemoney_plus",
            "avg_of_avg_handicap_plus", "ae_win_plus", "ae_place_plus", "uniq_win_ratio_plus",
            "avg_race_adjusted_residual_plus",  # NEW
        ]:
            summary[col] = np.nan
        return summary

    g_pos  = df_raw["pos_perc"].mean()
    g_prz  = df_raw["prizemoney"].mean()
    g_hcap = df_raw["handicapRatingKg"].mean()
    g_rar  = df_raw["race_adjusted_residual"].mean()  # NEW

    ae_mask = (
        df_raw["liveOdd"].notnull()
        & df_raw["odds_sum"].ge(1.1)
        & df_raw["odds_sum"].le(1.4)
    )
    ae_df = df_raw[ae_mask]
    if not ae_df.empty:
        g_ae_win = (ae_df["win"] / ae_df["odds"]).replace([np.inf, np.nan], 0).mean()
        g_ae_place = (ae_df["place"] / ae_df["odds_place"]).replace([np.inf, np.nan], 0).mean()
    else:
        g_ae_win = np.nan
        g_ae_place = np.nan

    unique_runners = df_raw.groupby(group_by)["horseId"].nunique()
    unique_winners = df_raw[df_raw["win"] == 1].groupby(group_by)["horseId"].nunique()
    total_runners = unique_runners.sum()
    total_winners = unique_winners.sum()
    g_uniq = (total_winners / total_runners) if total_runners > 0 else np.nan

    summary["avg_pos_perc_plus"]              = (summary["avg_pos_perc"]              / g_pos      * 100).round(1)
    summary["avg_prizemoney_plus"]            = (summary["avg_prizemoney"]            / g_prz      * 100).round(1)
    summary["avg_of_avg_handicap_plus"]       = (summary["avg_of_avg_handicap"]       / g_hcap     * 100).round(1)
    summary["ae_win_plus"]                    = (summary["ae_win"]                    / g_ae_win   * 100).round(1)
    summary["ae_place_plus"]                  = (summary["ae_place"]                  / g_ae_place * 100).round(1)
    summary["uniq_win_ratio_plus"]            = (summary["uniq_win_ratio"]            / g_uniq     * 100).round(1)
    summary["avg_race_adjusted_residual_plus"] = (summary["avg_race_adjusted_residual"] / g_rar    * 100).round(1)  # NEW

    return summary


def compute_general_stats(
    runners: pd.DataFrame,
    group_by: str = "horseSir",
    min_runs: int = 100,
    top_n: int = 15000,
    filter_feature: str = None,
    filter_value=None,
    filter_time: str = None,
    date_column: str = "date",
    YoY: bool = False
) -> pd.DataFrame:
    df = runners.copy()

    if filter_feature and filter_value is not None:
        if filter_feature not in df.columns:
            raise KeyError(f"Column '{filter_feature}' not found")
        df = df[df[filter_feature] == filter_value]

    if date_column not in df.columns:
        raise KeyError(f"Date column '{date_column}' not found in DataFrame")
    if not np.issubdtype(df[date_column].dtype, np.datetime64):
        df[date_column] = pd.to_datetime(df[date_column], errors='coerce')

    if filter_time:
        delta = pd.Timedelta(filter_time)
        now = pd.Timestamp.now()
        cutoff_current = now - delta
        df_current = df[df[date_column] >= cutoff_current]
    else:
        df_current = df.copy()

    summary_current = _build_summary(df_current, group_by)
    summary_current = _add_ops(summary_current, df_current, group_by=group_by, min_runs=min_runs)

    cols = [
        group_by, "runs", "wins", "win_pct", "ae_win", "ae_win_plus",
        "places", "place_pct", "ae_place", "ae_place_plus",
        "unique_runners", "unique_winners", "uniq_win_ratio", "uniq_win_ratio_plus",
        "avg_pos_perc", "avg_pos_perc_plus",
        "avg_prizemoney", "avg_prizemoney_plus",
        "avg_of_avg_handicap", "avg_of_avg_handicap_plus",
        "avg_race_adjusted_residual", "avg_race_adjusted_residual_plus",
        "std_race_adjusted_residual"   # NEW
    ]


    result = summary_current[cols].sort_values("wins", ascending=False).head(top_n)

    if YoY and filter_time:
        cutoff_prior_start = now - delta * 2
        cutoff_prior_end = now - delta
        df_prior = df[(df[date_column] >= cutoff_prior_start) & (df[date_column] < cutoff_prior_end)]

        summary_prior = _build_summary(df_prior, group_by)
        summary_prior = _add_ops(summary_prior, df_prior, group_by=group_by, min_runs=min_runs)
        summary_prior = summary_prior[cols]

        merged = result.merge(
            summary_prior,
            on=group_by,
            how="left",
            suffixes=("", "_prior")
        )

        for col in cols[1:]:
            prior_col = f"{col}_prior"
            delta_col = f"{col}_YoY"
            merged[delta_col] = merged[col] - merged[prior_col]

        ordered_cols = [group_by]
        for col in cols[1:]:
            ordered_cols.append(col)
            ordered_cols.append(f"{col}_YoY")

        return merged[ordered_cols].sort_values("wins", ascending=False).head(top_n)

    return result
```

```python
def compute_race_stats(runners_tdy, runners, weights=None, race_idx=11):
    """
    Compute trainer, jockey, trainer-jockey, horseSir, and owner stats for a specific race.
    """
    import pandas as pd

    race_id = runners_tdy.raceId.unique()[race_idx]
    race_runners = runners_tdy[runners_tdy.raceId == race_id].copy()
    meeting_name = race_runners['meetingName'].iloc[0]

    def get_stats(group_by, filter_feature=None, filter_value=None, period="30D", col_suffix="", yoy=True):
        df = compute_general_stats(
            runners, group_by=group_by,
            filter_feature=filter_feature, filter_value=filter_value,
            filter_time=period, YoY=yoy
        ).loc[:, [group_by, "avg_race_adjusted_residual_plus"]]
        return df.rename(columns={"avg_race_adjusted_residual_plus": col_suffix})

    # Trainer stats
    stats_t_30  = get_stats("trainerName", period="30D",  col_suffix="ou_30D_t")
    stats_t_365 = get_stats("trainerName", period="365D", col_suffix="ou_365D_t")
    stats_t_750_meeting = get_stats("trainerName", "meetingName", meeting_name, "750D", col_suffix="ou_750D_tr_meeting")

    race_type = race_runners['raceType'].iloc[0]
    stats_t_365_raceType = get_stats(
        "trainerName", "type", race_type, "365D", col_suffix="ou_365D_t_raceType"
    )

    # Jockey stats
    stats_j_30  = get_stats("jockeyName", period="30D",  col_suffix="ou_30D_j")
    stats_j_365 = get_stats("jockeyName", period="365D", col_suffix="ou_365D_j")
    stats_j_750_meeting = get_stats("jockeyName", "meetingName", meeting_name, "750D", col_suffix="ou_750D_j_meeting")

    # Trainer-Jockey stats
    tj_pairs = race_runners[["trainerName", "jockeyName"]].drop_duplicates()
    tj_stats_list = []
    for _, row in tj_pairs.iterrows():
        trainer, jockey = row["trainerName"], row["jockeyName"]
        tj_stats = compute_general_stats(
            runners, group_by="trainerName", filter_feature="jockeyName",
            filter_value=jockey, filter_time="365D", YoY=False
        )
        tj_stats = (
            tj_stats[tj_stats["trainerName"] == trainer]
            .loc[:, ["trainerName", "avg_race_adjusted_residual_plus"]]
            .rename(columns={"avg_race_adjusted_residual_plus": "ou_365D_tj"})
        )
        tj_stats["jockeyName"] = jockey
        tj_stats_list.append(tj_stats)

    stats_tj_365 = pd.concat(tj_stats_list, ignore_index=True)

    # HorseSir stats
    distance_group = race_runners['distance_group'].iloc[0]
    going_category = race_runners['going_category'].iloc[0]

    stats_hs = get_stats(
        "horseSir", period="9999D", col_suffix="ou_9999D_hs", yoy=False
    )
    stats_hs_dis = get_stats(
        "horseSir", filter_feature="distance_group", filter_value=distance_group,
        period="9999D", col_suffix="ou_9999D_hs_dis", yoy=False
    )
    stats_hs_going = get_stats(
        "horseSir", filter_feature="going_category", filter_value=going_category,
        period="9999D", col_suffix="ou_9999D_hs_going", yoy=False
    )

    # Horse stats
    stats_h = get_stats(
        "horseName", period="365D", col_suffix="ou_365D_h", yoy=False
    )
    stats_h_going = get_stats(
        "horseName", filter_feature="going_category", filter_value=going_category,
        period="750D", col_suffix="ou_750D_h_going", yoy=False
    )
    stats_h_dis = get_stats(
        "horseName", filter_feature="distance_group", filter_value=distance_group,
        period="750D", col_suffix="ou_750D_h_dis", yoy=False
    )


    # Owner stats
    stats_own_365 = get_stats(
        "ownerName", period="365D", col_suffix="ou_365D_own"
    )

    # Merge all stats
    result = race_runners \
        .merge(stats_t_30, on="trainerName", how="left") \
        .merge(stats_t_365, on="trainerName", how="left") \
        .merge(stats_t_750_meeting, on="trainerName", how="left") \
        .merge(stats_t_365_raceType, on="trainerName", how="left") \
        .merge(stats_j_30, on="jockeyName", how="left") \
        .merge(stats_j_365, on="jockeyName", how="left") \
        .merge(stats_h, on="horseName", how="left") \
        .merge(stats_h_going, on="horseName", how="left") \
        .merge(stats_h_dis, on="horseName", how="left") \
        .merge(stats_j_750_meeting, on="jockeyName", how="left") \
        .merge(stats_tj_365, on=["trainerName", "jockeyName"], how="left") \
        .merge(stats_hs, on="horseSir", how="left") \
        .merge(stats_hs_dis, on="horseSir", how="left") \
        .merge(stats_hs_going, on="horseSir", how="left") \
        .merge(stats_own_365, on="ownerName", how="left")

    # Select and rename
    final_cols = {
        "horseName": "horseName",
        "horseSir": "horseSir",
        "trainerName": "trainerName",
        "jockeyName": "jockeyName",
        "ownerName": "ownerName",
        "ou_30D_t": "ou_30D_t",
        "ou_365D_t": "ou_365D_t",
        "ou_30D_j": "ou_30D_j",
        "ou_365D_j": "ou_365D_j",
        "ou_750D_tr_meeting": "ou_750D_tr_meeting",
        "ou_750D_j_meeting": "ou_750D_j_meeting",
        "ou_365D_tj": "ou_365D_tj",
        "ou_365D_t_raceType": "ou_365D_t_raceType",
        "ou_9999D_hs": "ou_9999D_hs",
        "ou_9999D_hs_dis": "ou_9999D_hs_dis",
        "ou_9999D_hs_going": "ou_9999D_hs_going",
        "ou_365D_own": "ou_365D_own",
        "ou_365D_h": "ou_365D_h",
        "ou_750D_h_going": "ou_750D_h_going",
        "ou_750D_h_dis": "ou_750D_h_dis"
    }

    final_df = result[list(final_cols.keys())].rename(columns=final_cols).fillna(90)

    # Median & mean stats
    stat_cols = [
        "ou_30D_t", "ou_365D_t", "ou_30D_j", "ou_365D_j",
        "ou_750D_tr_meeting", "ou_750D_j_meeting", "ou_365D_tj", "ou_365D_t_raceType",
        "ou_9999D_hs", "ou_9999D_hs_dis", "ou_9999D_hs_going", "ou_365D_own",
        "ou_365D_h", "ou_750D_h_going", "ou_750D_h_dis"
    ]


    final_df["median_stat"] = final_df[stat_cols].median(axis=1).round(1)
    final_df["mean_stat"] = final_df[stat_cols].mean(axis=1).round(1)

    if weights is None:
        weights = {col: 1/len(stat_cols) for col in stat_cols}  # equal weights

    # Ensure weights sum to 1 if desired
    total_weight = sum(weights.values())
    weights = {k: v/total_weight for k, v in weights.items()}

    # Compute weighted score
    final_df["overall_score"] = sum(
        final_df[col] * weights.get(col, 0) for col in stat_cols
    ).round(1)


    return final_df.fillna(100).sort_values(by="overall_score", ascending=False).reset_index(drop=True)
```

```python
def full_history_dataframe(race_index, runners_tdy, df_with_ratings):
    """
    Return a DataFrame with full history of (smoothed, weight-adjusted) ratings for all horses in the selected race.

    Parameters:
        race_index (int): index into unique race IDs
        runners_tdy (DataFrame): runner metadata including 'raceId', 'horseId', 'horseName', 'weightKg'
        df_with_ratings (DataFrame): historical ratings with columns ['horseId', 'date', 'rating_after_race']

    Returns:
        DataFrame with columns: ['horseId', 'horseName', 'date', 'adjusted_rating_after_race']
    """
    race_ids = runners_tdy['raceId'].unique()
    if race_index < 0 or race_index >= len(race_ids):
        raise IndexError(f"race_index {race_index} out of range (0 to {len(race_ids)-1})")
    race_id = race_ids[race_index]
    race_horses = (
        runners_tdy
        .loc[runners_tdy['raceId'] == race_id, ['horseId','horseName','weightKg']]
        .set_index('horseId')
    )
    horse_ids = race_horses.index

    df = df_with_ratings.copy()
    df['date'] = pd.to_datetime(df['date'])
    df = df[df['horseId'].isin(horse_ids)]
    df.sort_values(['horseId','date'], inplace=True)

    # Collect results here
    out = []
    for hid, grp in df.groupby('horseId'):
        wt_adj = (race_horses.at[hid, 'weightKg'] - 55) * 0.625
        series_full, _ = smoothed_series(grp, wt_adj, frac=0.4)
        df_tmp = pd.DataFrame({
            'horseId': hid,
            'horseName': race_horses.at[hid, 'horseName'],
            'date': series_full.index,
            'adjusted_rating_after_race': series_full.values
        })
        out.append(df_tmp)
    result_df = pd.concat(out, ignore_index=True)
    return result_df
```

```python
import pandas as pd
from scipy.stats import ttest_ind

def compute_preference_stats(runners, runners_tdy):
    """
    Compute statistics & t-tests for horses in runners_tdy against the rest of the data.
    Suppresses t-tests if either group has < 5 observations.
    """
    results = []

    for _, tdy_row in runners_tdy.iterrows():
        horseId = tdy_row['horseId']
        horseName = tdy_row['horseName']
        raceId = tdy_row['raceId']
        going_cat = tdy_row['going_category']
        dist_grp = tdy_row['distance_group']
        meet_name = tdy_row['meetingName']

        row_stats = {
            'horseId': horseId,
            'horseName': horseName,
            'raceId': raceId
        }

        def aggregate(group):
            return pd.Series({
                'mean_pos': group['pos_perc'].mean(),
                'runs': len(group),
                'wins': group['win'].sum(),
                'places': group['place'].sum()
            })

        categories = {
            'overall': runners[runners['horseId'] == horseId],
            'going_category': runners[(runners['horseId'] == horseId) & (runners['going_category'] == going_cat)],
            'distance_group': runners[(runners['horseId'] == horseId) & (runners['distance_group'] == dist_grp)],
            'meetingName': runners[(runners['horseId'] == horseId) & (runners['meetingName'] == meet_name)],
        }

        for cat, group in categories.items():
            agg = aggregate(group)
            row_stats[f'{cat}_mean'] = agg['mean_pos']
            row_stats[f'{cat}_runs'] = agg['runs']
            row_stats[f'{cat}_wins'] = agg['wins']
            row_stats[f'{cat}_places'] = agg['places']

            # t-test: this category vs all *other* categories
            if cat != 'overall':
                other_group = runners[(runners['horseId'] == horseId) & (runners[cat] != tdy_row[cat])]
            else:
                other_group = runners[runners['horseId'] == horseId]

            g_vals = group['pos_perc'].dropna()
            o_vals = other_group['pos_perc'].dropna()

            # only perform t-test if both groups have at least 5 observations,
            # at least 2 unique values each, and at least 2 observations each
            if (
                len(g_vals) >= 5 and len(o_vals) >= 5 and
                g_vals.nunique() > 1 and o_vals.nunique() > 1
            ):
                tscore, pval = ttest_ind(g_vals, o_vals, equal_var=False)
            else:
                tscore, pval = None, None

            row_stats[f'{cat}_tscore'] = tscore
            row_stats[f'{cat}_pval'] = pval

        results.append(row_stats)

    df_results = pd.DataFrame(results)
    return df_results
```

```python
def parse_race_time(time_str):
    """
    Convert race time strings to decimal seconds.
    Handles formats like:
    - 2'18"2 (2 minutes, 18.2 seconds)
    - '56"33 (56.33 seconds)
    """
    if pd.isna(time_str):
        return None

    # Convert to string and strip whitespace
    time_str = str(time_str).strip()

    # Pattern for minutes'seconds"hundredths (e.g., 2'18"2)
    pattern_with_minutes = r"(\d+)'(\d+)\"(\d+)"

    # Pattern for 'seconds"hundredths (e.g., '56"33)
    pattern_seconds_only = r"'(\d+)\"(\d+)"

    # Try matching with minutes first
    match = re.match(pattern_with_minutes, time_str)
    if match:
        minutes = int(match.group(1))
        seconds = int(match.group(2))
        hundredths = int(match.group(3))
        # Convert to total seconds
        # Assuming hundredths is 2 digits (e.g., 2 means 0.2, 20 means 0.20)
        if len(match.group(3)) == 1:
            decimal_part = hundredths / 10
        else:
            decimal_part = hundredths / 100
        return minutes * 60 + seconds + decimal_part

    # Try matching seconds only
    match = re.match(pattern_seconds_only, time_str)
    if match:
        seconds = int(match.group(1))
        hundredths = int(match.group(2))
        # Convert hundredths to decimal
        if len(match.group(2)) == 1:
            decimal_part = hundredths / 10
        else:
            decimal_part = hundredths / 100
        return seconds + decimal_part

    # If no pattern matches, return None
    print(f"Warning: Could not parse time string: {time_str}")
    return None
```

```python
def summarize_trainer_stats_by_race(
    runners_tdy_df: pd.DataFrame,
    races_tdy_df: pd.DataFrame,
    runners_df: pd.DataFrame,
    race: int
) -> pd.DataFrame:
    """
    Summarize trainer stats (last 365 days up to race date) for all runners in a given race.
    Returns one row per horse in today's race with trainer stats attached.

    Columns returned:
    horseId, trainerName, runs, rtr, rtr_last20, rtr_raceType,
    rtr_meeting, rtr_trainer_jockey, rtr_trainer_owner

    Notes on rtr variants:
    - rtr: Mean rating improvement over last 365 days (all trainer runs)
    - rtr_last20: Mean rating improvement over last 20 runs (all trainer runs)
    - rtr_raceType: Mean rating improvement for same race type over last 750 days (all trainer runs)
    - rtr_meeting: Mean rating improvement at same meeting over last 750 days (all trainer runs)
    - rtr_trainer_jockey: Mean rating improvement for trainer-jockey combination over last 750 days
    - rtr_trainer_owner: Mean rating improvement for trainer-owner combination over last 750 days
    """

    # --- Identify race and date ---
    race_id = runners_tdy_df.raceId.unique()[race]
    print(f"Trainer Stats Table:")
    if races_tdy_df is not None:
        race_date = pd.to_datetime(races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'date'].iloc[0])
        race_type = races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'type'].iloc[0]
    else:
        raise ValueError("races_tdy_df is required to get the race date and type")

    cutoff_date_365 = race_date - pd.Timedelta(days=365)
    cutoff_date_750 = race_date - pd.Timedelta(days=750)

    # --- Get today's race info with meeting, jockey, and owner ---
    per_horse_today = (runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id,
                                         ['horseId', 'trainerName', 'meetingName', 'jockeyName', 'ownerName']]
                      .drop_duplicates())

    # --- Prepare historical data & compute per-run rating deltas ---
    hist = runners_df.copy()
    hist['date'] = pd.to_datetime(hist['date'], errors='coerce')
    hist = hist.sort_values(['horseId', 'date'])

    # Per horse, change in rating_after_race compared to previous start
    hist['run_to_rating'] = (hist
                            .groupby('horseId', group_keys=False)['rating_after_race']
                            .apply(lambda s: s.diff()))

    # Historical data within 750-day window (for specific metrics)
    historical_data_750 = hist[
        (hist['date'] >= cutoff_date_750) &
        (hist['date'] < race_date)
    ].copy()

    # --- Calculate all metrics for each horse ---
    horse_metrics = []

    for _, horse_row in per_horse_today.iterrows():
        horse_id = horse_row['horseId']
        trainer_name = horse_row['trainerName']
        meeting_name = horse_row['meetingName']
        jockey_name = horse_row['jockeyName']
        owner_name = horse_row['ownerName']

        # Get all historical data for this trainer (before race date only)
        trainer_hist = hist[
            (hist['trainerName'] == trainer_name) &
            (hist['date'] < race_date)
        ].copy()

        # Get trainer data within 365-day window
        trainer_365_data = trainer_hist[
            trainer_hist['date'] >= cutoff_date_365
        ].copy()

        # --- Calculate trainer stats (365 days) ---
        if trainer_365_data.empty:
            runs = 0
            rtr = 0.0
        else:
            runs = len(trainer_365_data)

            # rtr mean (365 days, all trainer runs)
            rtr = trainer_365_data['run_to_rating'].dropna().mean()
            if pd.isna(rtr):
                rtr = 0.0

        # Format rtr with runs count
        rtr_formatted = str(np.round(rtr, 2)) + ' (' + str(runs) + ')'

        # --- Calculate additional rtr metrics ---

        # 1. Last 20 runs of this trainer (across all horses, before race date)
        last20_data = trainer_hist.tail(20)
        rtr_last20 = last20_data['run_to_rating'].dropna().mean() if not last20_data.empty else 0.0

        # 2. Same race type for this trainer (within 750 days)
        if race_type is not None:
            type_data = historical_data_750[
                (historical_data_750['trainerName'] == trainer_name) &
                (historical_data_750['type'] == race_type)
            ]
            type_runs = len(type_data['run_to_rating'].dropna()) if not type_data.empty else 0
            rtr_raceType = type_data['run_to_rating'].dropna().mean() if not type_data.empty else 0.0
        else:
            type_runs = 0
            rtr_raceType = 0.0

        # 3. Same meeting for this trainer (within 750 days)
        meeting_data = historical_data_750[
            (historical_data_750['trainerName'] == trainer_name) &
            (historical_data_750['meetingName'] == meeting_name)
        ]
        meeting_runs = len(meeting_data['run_to_rating'].dropna()) if not meeting_data.empty else 0
        rtr_meeting = meeting_data['run_to_rating'].dropna().mean() if not meeting_data.empty else 0.0

        # 4. Trainer-jockey combination (within 750 days)
        trainer_jockey_data = historical_data_750[
            (historical_data_750['trainerName'] == trainer_name) &
            (historical_data_750['jockeyName'] == jockey_name)
        ]
        trainer_jockey_runs = len(trainer_jockey_data['run_to_rating'].dropna()) if not trainer_jockey_data.empty else 0
        rtr_trainer_jockey = trainer_jockey_data['run_to_rating'].dropna().mean() if not trainer_jockey_data.empty else 0.0

        # 5. Trainer-owner combination (within 750 days)
        trainer_owner_data = historical_data_750[
            (historical_data_750['trainerName'] == trainer_name) &
            (historical_data_750['ownerName'] == owner_name)
        ]
        trainer_owner_runs = len(trainer_owner_data['run_to_rating'].dropna()) if not trainer_owner_data.empty else 0
        rtr_trainer_owner = trainer_owner_data['run_to_rating'].dropna().mean() if not trainer_owner_data.empty else 0.0

        horse_metrics.append({
            'horseId': horse_id,
            'trainerName': trainer_name,
            'runs': runs,
            'rtr': rtr_formatted,
            'rtr_last20': rtr_last20,
            'rtr_raceType': rtr_raceType,
            'rtr_meeting': rtr_meeting,
            'rtr_trainer_jockey': rtr_trainer_jockey,
            'rtr_trainer_owner': rtr_trainer_owner,
            'type_runs': type_runs,
            'meeting_runs': meeting_runs,
            'trainer_jockey_runs': trainer_jockey_runs,
            'trainer_owner_runs': trainer_owner_runs
        })

    horse_metrics_df = pd.DataFrame(horse_metrics)

    # Round and format rtr metrics
    horse_metrics_df['rtr_last20'] = np.round(horse_metrics_df['rtr_last20'].fillna(0.0), 2)

    # Format the four metrics with sample sizes in brackets
    horse_metrics_df['rtr_raceType'] = (
        np.round(horse_metrics_df['rtr_raceType'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['type_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_meeting'] = (
        np.round(horse_metrics_df['rtr_meeting'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['meeting_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_trainer_jockey'] = (
        np.round(horse_metrics_df['rtr_trainer_jockey'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['trainer_jockey_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_trainer_owner'] = (
        np.round(horse_metrics_df['rtr_trainer_owner'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['trainer_owner_runs'].astype(str) + ')'
    )

    # --- Prepare final output ---
    out = (per_horse_today
           .merge(horse_metrics_df[['horseId', 'runs', 'rtr', 'rtr_last20',
                                   'rtr_raceType', 'rtr_meeting',
                                   'rtr_trainer_jockey', 'rtr_trainer_owner']],
                  on='horseId', how='left')
           .fillna({
               'runs': 0, 'rtr': '0.0 (0)',
               'rtr_last20': 0.0, 'rtr_raceType': '0.0 (0)',
               'rtr_meeting': '0.0 (0)', 'rtr_trainer_jockey': '0.0 (0)',
               'rtr_trainer_owner': '0.0 (0)'
           }))

    # --- Order & dtypes ---
    out = out[['horseId', 'trainerName', 'rtr', 'rtr_last20',
               'rtr_raceType', 'rtr_meeting', 'rtr_trainer_jockey',
               'rtr_trainer_owner']].sort_values(by='rtr', ascending=False)

    out['rtr'] = out['rtr'].astype(str)

    return out

def summarize_sire_stats_by_race(
    runners_tdy_df: pd.DataFrame,
    races_tdy_df: pd.DataFrame,
    runners_df: pd.DataFrame,
    race: int
) -> pd.DataFrame:
    """
    Summarize sire stats (all historical data up to race date) for all runners in a given race.
    Returns one row per horse in today's race with sire stats attached.

    Columns returned:
    horseId, horseSir, runs, rtr, rtr_going_category,
    rtr_distance_group, rtr_trainer_sire

    Notes on rtr variants:
    - rtr: Mean rating improvement over all historical data (all sire progeny)
    - rtr_going_category: Mean rating improvement for same going category (all sire progeny)
    - rtr_distance_group: Mean rating improvement for same distance group (all sire progeny)
    - rtr_trainer_sire: Mean rating improvement for trainer-sire combination
    """

    # --- Identify race and date ---
    race_id = runners_tdy_df.raceId.unique()[race]
    print(f"Sire Stats Table:")
    if races_tdy_df is not None:
        race_date = pd.to_datetime(races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'date'].iloc[0])
    else:
        raise ValueError("races_tdy_df is required to get the race date")

    # Get going_category and distance_group from runners_tdy_df for this race
    race_info = runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id, ['going_category', 'distance_group']].iloc[0]
    going_category = race_info['going_category']
    distance_group = race_info['distance_group']

    # --- Get today's race info with trainer and sire ---
    per_horse_today = (runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id,
                                         ['horseId', 'horseSir', 'trainerName']]
                      .drop_duplicates())

    # --- Prepare historical data & compute per-run rating deltas ---
    hist = runners_df.copy()
    hist['date'] = pd.to_datetime(hist['date'], errors='coerce')
    hist = hist.sort_values(['horseId', 'date'])

    # Per horse, change in rating_after_race compared to previous start
    hist['run_to_rating'] = (hist
                            .groupby('horseId', group_keys=False)['rating_after_race']
                            .apply(lambda s: s.diff()))

    # Historical data before race date
    historical_data = hist[hist['date'] < race_date].copy()

    # --- Calculate all metrics for each horse ---
    horse_metrics = []

    for _, horse_row in per_horse_today.iterrows():
        horse_id = horse_row['horseId']
        sire_name = horse_row['horseSir']
        trainer_name = horse_row['trainerName']

        # Get all historical data for this sire (before race date only)
        sire_hist = historical_data[
            historical_data['horseSir'] == sire_name
        ].copy()

        # --- Calculate sire stats (all historical data) ---
        if sire_hist.empty:
            runs = 0
            rtr = 0.0
        else:
            runs = len(sire_hist)

            # rtr mean (all historical data, all sire progeny)
            rtr = sire_hist['run_to_rating'].dropna().mean()
            if pd.isna(rtr):
                rtr = 0.0

        # Format rtr with runs count
        rtr_formatted = str(np.round(rtr, 2)) + ' (' + str(runs) + ')'

        # --- Calculate additional rtr metrics ---

        # 1. Same going category for this sire
        if going_category is not None:
            going_data = historical_data[
                (historical_data['horseSir'] == sire_name) &
                (historical_data['going_category'] == going_category)
            ]
            going_runs = len(going_data['run_to_rating'].dropna()) if not going_data.empty else 0
            rtr_going_category = going_data['run_to_rating'].dropna().mean() if not going_data.empty else 0.0
        else:
            going_runs = 0
            rtr_going_category = 0.0

        # 2. Same distance group for this sire
        if distance_group is not None:
            distance_data = historical_data[
                (historical_data['horseSir'] == sire_name) &
                (historical_data['distance_group'] == distance_group)
            ]
            distance_runs = len(distance_data['run_to_rating'].dropna()) if not distance_data.empty else 0
            rtr_distance_group = distance_data['run_to_rating'].dropna().mean() if not distance_data.empty else 0.0
        else:
            distance_runs = 0
            rtr_distance_group = 0.0

        # 3. Trainer-sire combination
        trainer_sire_data = historical_data[
            (historical_data['trainerName'] == trainer_name) &
            (historical_data['horseSir'] == sire_name)
        ]
        trainer_sire_runs = len(trainer_sire_data['run_to_rating'].dropna()) if not trainer_sire_data.empty else 0
        rtr_trainer_sire = trainer_sire_data['run_to_rating'].dropna().mean() if not trainer_sire_data.empty else 0.0

        horse_metrics.append({
            'horseId': horse_id,
            'horseSir': sire_name,
            'runs': runs,
            'rtr': rtr_formatted,
            'rtr_going_category': rtr_going_category,
            'rtr_distance_group': rtr_distance_group,
            'rtr_trainer_sire': rtr_trainer_sire,
            'going_runs': going_runs,
            'distance_runs': distance_runs,
            'trainer_sire_runs': trainer_sire_runs
        })

    horse_metrics_df = pd.DataFrame(horse_metrics)

    # Format the metrics with sample sizes in brackets
    horse_metrics_df['rtr_going_category'] = (
        np.round(horse_metrics_df['rtr_going_category'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['going_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_distance_group'] = (
        np.round(horse_metrics_df['rtr_distance_group'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['distance_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_trainer_sire'] = (
        np.round(horse_metrics_df['rtr_trainer_sire'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['trainer_sire_runs'].astype(str) + ')'
    )

    # --- Prepare final output ---
    out = (per_horse_today
           .merge(horse_metrics_df[['horseId', 'runs', 'rtr',
                                   'rtr_going_category', 'rtr_distance_group',
                                   'rtr_trainer_sire']],
                  on='horseId', how='left')
           .fillna({
               'runs': 0, 'rtr': '0.0 (0)',
               'rtr_going_category': '0.0 (0)',
               'rtr_distance_group': '0.0 (0)', 'rtr_trainer_sire': '0.0 (0)'
           }))

    # --- Order & dtypes ---
    out = out[['horseId', 'horseSir', 'rtr',
               'rtr_going_category', 'rtr_distance_group']].sort_values(by='rtr', ascending=False)

    out['rtr'] = out['rtr'].astype(str)

    return out


def summarize_draw_stats_by_race(
    runners_tdy_df: pd.DataFrame,
    races_tdy_df: pd.DataFrame,
    runners_df: pd.DataFrame,
    race: int
) -> pd.DataFrame:
    """
    Summarize draw stats (last 750 days up to race date) for all runners in a given race.
    Returns one row per horse in today's race with draw stats attached.

    Columns returned:
    horseId, draw, rtr

    Notes on rtr:
    - rtr: Mean rating improvement over last 750 days for this draw position
    """

    # --- Identify race and date ---
    race_id = runners_tdy_df.raceId.unique()[race]
    print(f"Draw Stats Table:")
    if races_tdy_df is not None:
        race_date = pd.to_datetime(races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'date'].iloc[0])
    else:
        raise ValueError("races_tdy_df is required to get the race date")

    cutoff_date_750 = race_date - pd.Timedelta(days=9999)

    # --- Get today's race info with draw_unique ---
    per_horse_today = (runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id,
                                         ['horseId', 'draw_unique']]
                      .drop_duplicates())

    # Extract just the draw number from draw_unique (first part before " - ")
    per_horse_today['draw'] = per_horse_today['draw_unique'].str.split(' - ').str[0]

    # --- Prepare historical data & compute per-run rating deltas ---
    hist = runners_df.copy()
    hist['date'] = pd.to_datetime(hist['date'], errors='coerce')
    hist = hist.sort_values(['horseId', 'date'])

    # Per horse, change in rating_after_race compared to previous start
    hist['run_to_rating'] = (hist
                            .groupby('horseId', group_keys=False)['rating_after_race']
                            .apply(lambda s: s.diff()))

    # Historical data within 750-day window
    historical_data_750 = hist[
        (hist['date'] >= cutoff_date_750) &
        (hist['date'] < race_date)
    ].copy()

    # --- Calculate draw metrics for each horse ---
    horse_metrics = []

    for _, horse_row in per_horse_today.iterrows():
        horse_id = horse_row['horseId']
        draw_unique = horse_row['draw_unique']
        draw = horse_row['draw']

        # --- Calculate draw stats (750 days) ---
        # All runs from this draw position within 750 days (using draw_unique for calculations)
        draw_data = historical_data_750[
            historical_data_750['draw_unique'] == draw_unique
        ]
        draw_runs = len(draw_data['run_to_rating'].dropna()) if not draw_data.empty else 0
        rtr = draw_data['run_to_rating'].dropna().mean() if not draw_data.empty else 0.0

        horse_metrics.append({
            'horseId': horse_id,
            'draw': draw,
            'rtr': rtr,
            'draw_runs': draw_runs
        })

    horse_metrics_df = pd.DataFrame(horse_metrics)

    # Format the metrics with sample sizes in brackets
    horse_metrics_df['rtr'] = (
        np.round(horse_metrics_df['rtr'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['draw_runs'].astype(str) + ')'
    )

    # --- Prepare final output ---
    out = (per_horse_today
           .merge(horse_metrics_df[['horseId', 'rtr']],
                  on='horseId', how='left')
           .fillna({
               'rtr': '0.0 (0)'
           }))

    # --- Order & dtypes ---
    out = out[['horseId', 'draw', 'rtr']].sort_values(by='rtr', ascending=False)

    out['rtr'] = out['rtr'].astype(str)

    return out

def get_horse_form_table(
    runners_tdy_df: pd.DataFrame,
    races_tdy_df: pd.DataFrame,
    runners_df: pd.DataFrame,
    race: int
) -> pd.DataFrame:
    """
    Get form table showing last 5 races for each runner in the target race.
    Returns up to 5 rows per horse with detailed form information.

    Columns returned:
    horseId, horseName, date, meetingName, raceType, class, distance, going_category,
    raceTotalPrize, draw, adjusted_handicapRatingKg, hood, tongueTie, blinkers,
    ranking, runners, cumulative_lengths_back, comment, liveOdd,
    competing_horses_today, median_top_half_rating_change

    Notes:
    - adjusted_handicapRatingKg = handicapRatingKg + 55 - TODAY'S weightKg (from runners_tdy_df)
    - competing_horses_today shows horses from today's race that also ran in each historical race,
      with adjusted cumulative lengths back difference in brackets
    - median_top_half_rating_change = median rating change of horses that finished in top half
      of this historical race, measured from their last available race within next 3 races
    """

    # --- Identify race and date ---
    race_id = runners_tdy_df.raceId.unique()[race]
    print(f"Form Table:")

    if races_tdy_df is not None:
        race_date = pd.to_datetime(races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'date'].iloc[0])
    else:
        raise ValueError("races_tdy_df is required to get the race date")

    # --- Get today's race runners with their weights ---
    todays_runners = (runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id,
                                        ['horseId', 'horseName', 'weightKg']]
                     .drop_duplicates())

    # Create mapping for today's weights
    todays_weights = dict(zip(todays_runners['horseId'], todays_runners['weightKg']))
    todays_horse_names = dict(zip(todays_runners['horseId'], todays_runners['horseName']))

    # --- Prepare historical data ---
    hist = runners_df.copy()
    hist['date'] = pd.to_datetime(hist['date'], errors='coerce')
    hist = hist.sort_values(['horseId', 'date'])

    # Filter to races before the target race date
    historical_data = hist[hist['date'] < race_date].copy()

    # --- Get form data for each horse ---
    form_records = []

    for _, horse_row in todays_runners.iterrows():
        horse_id = horse_row['horseId']
        horse_name = horse_row['horseName']
        today_weight = horse_row['weightKg']

        # Get last 5 races for this horse
        horse_form = (historical_data[historical_data['horseId'] == horse_id]
                     #.tail(5)  # Get last 5 races
                     .copy())

        if horse_form.empty:
            continue

        # Calculate adjusted handicap rating using TODAY'S weight for this horse
        horse_form['adjusted_handicapRatingKg'] = (
            horse_form['handicapRatingKg'] + 55 - today_weight
        )

        # Calculate adjusted handicap rating using TODAY'S weight for this horse
        horse_form['adjusted_rar'] = (
            horse_form['rating_after_race'] + 55 - today_weight
        )

        # For each form race, find other horses from today's race that also ran
        for _, form_race in horse_form.iterrows():
            form_race_id = form_race['raceId']
            form_date = form_race['date']

            # Find all horses from today's race that ran in this historical race
            competing_today = []

            # Get all runners from this historical race
            historical_race_runners = historical_data[
                (historical_data['raceId'] == form_race_id) &
                (historical_data['date'] == form_date)
            ].copy()

            for other_horse_id in todays_runners['horseId']:
                if other_horse_id == horse_id:
                    continue  # Skip the current horse

                # Check if this other horse ran in the historical race
                other_horse_form = historical_race_runners[
                    historical_race_runners['horseId'] == other_horse_id
                ]

                if not other_horse_form.empty:
                    other_horse_record = other_horse_form.iloc[0]
                    other_horse_name = todays_horse_names[other_horse_id]

                    # Calculate adjusted cumulative lengths back difference
                    current_horse_clb = form_race['cumulative_lengths_back']
                    other_horse_clb = other_horse_record['cumulative_lengths_back']

                    # Base difference (negative means current horse was ahead)
                    clb_diff = current_horse_clb - other_horse_clb

                    # Weight adjustments
                    current_horse_past_weight = form_race['weightKg']
                    other_horse_past_weight = other_horse_record['weightKg']
                    current_horse_today_weight = today_weight  # Use today_weight here
                    other_horse_today_weight = todays_weights[other_horse_id]

                    # Adjust for past weight difference (heavier = disadvantage)
                    past_weight_adjustment = (current_horse_past_weight - other_horse_past_weight)

                    # Adjust for today's weight difference (heavier today = less advantage)
                    today_weight_adjustment = -(current_horse_today_weight - other_horse_today_weight)

                    # Final adjusted difference
                    adjusted_diff = clb_diff + past_weight_adjustment + today_weight_adjustment

                    # Format the competing horse entry
                    competing_today.append(f"{other_horse_name} ({adjusted_diff:.1f})")

            # Join competing horses
            competing_horses_str = ', '.join(competing_today) if competing_today else '-'

            # --- Calculate median rating change for top half finishers ---
            median_rating_change = None

            if not historical_race_runners.empty:
                # Determine top half threshold
                num_runners = form_race['runners']
                top_half_threshold = num_runners / 2

                # Get horses that finished in top half (excluding target horse)
                top_half_horses = historical_race_runners[
                    (historical_race_runners['horseId'] != horse_id)
                ].copy()

                rating_changes = []

                for _, top_half_horse in top_half_horses.iterrows():
                    top_half_horse_id = top_half_horse['horseId']
                    top_half_race_rating = top_half_horse['rating_after_race']

                    # Get subsequent races for this horse (up to 3 races after the form race)
                    subsequent_races = hist[
                        (hist['horseId'] == top_half_horse_id) &
                        (hist['date'] > form_date)
                    ].head(3)

                    # Take the last available race from the subsequent races
                    if not subsequent_races.empty:
                        last_subsequent_race = subsequent_races.iloc[-1]
                        last_rating = last_subsequent_race['rating_after_race']

                        # Calculate rating change
                        if pd.notna(top_half_race_rating) and pd.notna(last_rating):
                            rating_change = last_rating - top_half_race_rating
                            rating_changes.append(rating_change)

                # Calculate median if we have rating changes
                if rating_changes:
                    median_rating_change = pd.Series(rating_changes).median()

            # Create form record
            form_record = {
                'horseId': horse_id,
                'horseName': horse_name,
                'date': form_race['date'],
                'meetingName': form_race['meetingName'],
                'raceType': form_race['raceType'],
                'class': form_race['class'],
                'distance': form_race['distance'],
                'going_category': form_race['going_category'],
                'raceTotalPrize': form_race['raceTotalPrize'],
                'draw': form_race['draw'],
                'jockey': form_race['jockeyName'],
                'adjusted_handicapRatingKg': form_race['adjusted_handicapRatingKg'],
                'hood': form_race['hood'],
                'tongueTie': form_race['tongueTie'],
                'blinkers': form_race['blinkers'],
                'ranking': form_race['ranking'],
                'runners': form_race['runners'],
                'cumulative_lengths_back': form_race['cumulative_lengths_back'],
                'comment': form_race['comment'],
                'liveOdd': form_race['liveOdd'],
                'competing_horses_today': competing_horses_str,
                'adjusted_rar': form_race['adjusted_rar'],
                'median_top_half_rating_change': median_rating_change
            }

            form_records.append(form_record)

    # Convert to DataFrame
    form_df = pd.DataFrame(form_records)

    if form_df.empty:
        # Return empty dataframe with correct columns if no form data
        columns = ['horseId', 'horseName', 'date', 'meetingName', 'raceType', 'class',
                  'distance', 'going_category', 'raceTotalPrize', 'draw',
                  'adjusted_handicapRatingKg', 'hood', 'tongueTie', 'blinkers',
                  'ranking', 'runners', 'cumulative_lengths_back', 'comment',
                  'liveOdd', 'adjusted_rar', 'competing_horses_today',
                  'median_top_half_rating_change']
        return pd.DataFrame(columns=columns)

    # Sort by horse name and then by date (most recent first for each horse)
    form_df = form_df.sort_values(['horseName', 'date'], ascending=[True, False])

    return form_df


def summarize_jockey_stats_by_race(
    runners_tdy_df: pd.DataFrame,
    races_tdy_df: pd.DataFrame,
    runners_df: pd.DataFrame,
    race: int
) -> pd.DataFrame:
    """
    Summarize jockey stats (last 365 days up to race date) for all runners in a given race.
    Returns one row per horse in today's race with jockey stats attached.

    Columns returned:
    horseId, jockeyName, runs, rtr, rtr_last20, rtr_going_category,
    rtr_meeting, rtr_trainer_jockey

    Notes on rtr variants:
    - rtr: Mean rating improvement over last 365 days (all jockey runs)
    - rtr_last20: Mean rating improvement over last 20 runs (all jockey runs)
    - rtr_going_category: Mean rating improvement for same going category over last 750 days (all jockey runs)
    - rtr_meeting: Mean rating improvement at same meeting over last 750 days (all jockey runs)
    - rtr_trainer_jockey: Mean rating improvement for trainer-jockey combination over last 750 days
    """

    # --- Identify race and date ---
    race_id = runners_tdy_df.raceId.unique()[race]
    print(f"Jockey Stats Table:")
    if races_tdy_df is not None:
        race_date = pd.to_datetime(races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'date'].iloc[0])
    else:
        raise ValueError("races_tdy_df is required to get the race date")

    # Get going_category from runners_tdy_df for this race
    going_category = runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id, 'going_category'].iloc[0]

    cutoff_date_365 = race_date - pd.Timedelta(days=365)
    cutoff_date_750 = race_date - pd.Timedelta(days=750)

    # --- Get today's race info with meeting and trainer ---
    per_horse_today = (runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id,
                                         ['horseId', 'jockeyName', 'meetingName', 'trainerName']]
                      .drop_duplicates())

    # --- Prepare historical data & compute per-run rating deltas ---
    hist = runners_df.copy()
    hist['date'] = pd.to_datetime(hist['date'], errors='coerce')
    hist = hist.sort_values(['horseId', 'date'])

    # Per horse, change in rating_after_race compared to previous start
    hist['run_to_rating'] = (hist
                            .groupby('horseId', group_keys=False)['rating_after_race']
                            .apply(lambda s: s.diff()))

    # Historical data within 750-day window (for specific metrics)
    historical_data_750 = hist[
        (hist['date'] >= cutoff_date_750) &
        (hist['date'] < race_date)
    ].copy()

    # --- Calculate all metrics for each horse ---
    horse_metrics = []

    for _, horse_row in per_horse_today.iterrows():
        horse_id = horse_row['horseId']
        jockey_name = horse_row['jockeyName']
        meeting_name = horse_row['meetingName']
        trainer_name = horse_row['trainerName']

        # Get all historical data for this jockey (before race date only)
        jockey_hist = hist[
            (hist['jockeyName'] == jockey_name) &
            (hist['date'] < race_date)
        ].copy()

        # Get jockey data within 365-day window
        jockey_365_data = jockey_hist[
            jockey_hist['date'] >= cutoff_date_365
        ].copy()

        # --- Calculate jockey stats (365 days) ---
        if jockey_365_data.empty:
            runs = 0
            rtr = 0.0
        else:
            runs = len(jockey_365_data)

            # rtr mean (365 days, all jockey runs)
            rtr = jockey_365_data['run_to_rating'].dropna().mean()
            if pd.isna(rtr):
                rtr = 0.0

        # Format rtr with runs count
        rtr_formatted = str(np.round(rtr, 2)) + ' (' + str(runs) + ')'

        # --- Calculate additional rtr metrics ---

        # 1. Last 20 runs of this jockey (across all horses, before race date)
        last20_data = jockey_hist.tail(30)
        rtr_last20 = last20_data['run_to_rating'].dropna().mean() if not last20_data.empty else 0.0

        # 2. Same going category for this jockey (within 750 days)
        if going_category is not None:
            going_data = historical_data_750[
                (historical_data_750['jockeyName'] == jockey_name) &
                (historical_data_750['going_category'] == going_category)
            ]
            going_runs = len(going_data['run_to_rating'].dropna()) if not going_data.empty else 0
            rtr_going_category = going_data['run_to_rating'].dropna().mean() if not going_data.empty else 0.0
        else:
            going_runs = 0
            rtr_going_category = 0.0

        # 3. Same meeting for this jockey (within 750 days)
        meeting_data = historical_data_750[
            (historical_data_750['jockeyName'] == jockey_name) &
            (historical_data_750['meetingName'] == meeting_name)
        ]
        meeting_runs = len(meeting_data['run_to_rating'].dropna()) if not meeting_data.empty else 0
        rtr_meeting = meeting_data['run_to_rating'].dropna().mean() if not meeting_data.empty else 0.0

        # 4. Trainer-jockey combination (within 750 days)
        trainer_jockey_data = historical_data_750[
            (historical_data_750['trainerName'] == trainer_name) &
            (historical_data_750['jockeyName'] == jockey_name)
        ]
        trainer_jockey_runs = len(trainer_jockey_data['run_to_rating'].dropna()) if not trainer_jockey_data.empty else 0
        rtr_trainer_jockey = trainer_jockey_data['run_to_rating'].dropna().mean() if not trainer_jockey_data.empty else 0.0

        horse_metrics.append({
            'horseId': horse_id,
            'jockeyName': jockey_name,
            'runs': runs,
            'rtr': rtr_formatted,
            'rtr_last20': rtr_last20,
            'rtr_going_category': rtr_going_category,
            'rtr_meeting': rtr_meeting,
            'rtr_trainer_jockey': rtr_trainer_jockey,
            'going_runs': going_runs,
            'meeting_runs': meeting_runs,
            'trainer_jockey_runs': trainer_jockey_runs
        })

    horse_metrics_df = pd.DataFrame(horse_metrics)

    # Round and format rtr metrics
    horse_metrics_df['rtr_last20'] = np.round(horse_metrics_df['rtr_last20'].fillna(0.0), 2)

    # Format the three metrics with sample sizes in brackets
    horse_metrics_df['rtr_going_category'] = (
        np.round(horse_metrics_df['rtr_going_category'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['going_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_meeting'] = (
        np.round(horse_metrics_df['rtr_meeting'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['meeting_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_trainer_jockey'] = (
        np.round(horse_metrics_df['rtr_trainer_jockey'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['trainer_jockey_runs'].astype(str) + ')'
    )

    # --- Prepare final output ---
    out = (per_horse_today
           .merge(horse_metrics_df[['horseId', 'runs', 'rtr', 'rtr_last20',
                                   'rtr_going_category', 'rtr_meeting',
                                   'rtr_trainer_jockey']],
                  on='horseId', how='left')
           .fillna({
               'runs': 0, 'rtr': '0.0 (0)',
               'rtr_last20': 0.0, 'rtr_going_category': '0.0 (0)',
               'rtr_meeting': '0.0 (0)', 'rtr_trainer_jockey': '0.0 (0)'
           }))

    # --- Order & dtypes ---
    out = out[['horseId', 'jockeyName', 'rtr', 'rtr_last20',
               'rtr_going_category', 'rtr_meeting', 'rtr_trainer_jockey']].sort_values(by='rtr', ascending=False)

    out['rtr'] = out['rtr'].astype(str)

    return out

def summarize_horse_stats_by_race(
    runners_tdy_df: pd.DataFrame,
    races_tdy_df: pd.DataFrame,
    runners_df: pd.DataFrame,
    race: int
) -> pd.DataFrame:
    """
    Summarize horse stats (last 365 days up to race date) for all runners in a given race.
    Returns one row per horse in today's race with horse stats attached.

    Columns returned:
    horseName, sex, age, tongueTie, hood, blinkers, adjusted_handicapRatingKg,
    runs_wins_places_365d, runs_wins_places, rtr, rtr_going_category, rtr_distance_group,
    last_adjusted_rating, rtr_last_5, days_since_last_run

    Notes on rtr variants:
    - rtr: Mean rating improvement over last 365 days (this horse only)
    - rtr_going_category: Mean rating improvement for same going category (all time, this horse only)
    - rtr_distance_group: Mean rating improvement for same distance group (all time, this horse only)
    - rtr_last_5: Rating improvements for up to last 5 runs before race date
    - last_adjusted_rating: Last rating_after_race + 55 - TODAY'S weightKg (corrected)
    - days_since_last_run: Number of days between the most recent run and the target race date
    """

    # --- Identify race and date ---
    race_id = runners_tdy_df.raceId.unique()[race]
    print(f"Horse Stats Table:")
    if races_tdy_df is not None:
        race_date = pd.to_datetime(races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'date'].iloc[0])
        race_going_category = runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id, 'going_category'].iloc[0]
        race_distance_group = races_tdy_df.loc[races_tdy_df['id_race'] == race_id, 'distance_group'].iloc[0]
    else:
        raise ValueError("races_tdy_df is required to get the race date, going category, and distance group")

    cutoff_date_365 = race_date - pd.Timedelta(days=365)

    # --- Get today's race info ---
    per_horse_today = (runners_tdy_df.loc[runners_tdy_df['raceId'] == race_id,
                                         ['horseId', 'horseName', 'sex', 'age', 'tongueTie', 'hood', 'blinkers',
                                          'handicapRatingKg', 'weightKg']]
                      .drop_duplicates())

    # Calculate adjusted handicap rating
    per_horse_today['adjusted_handicapRatingKg'] = per_horse_today['handicapRatingKg'] - per_horse_today['weightKg'] + 55

    # --- Prepare historical data & compute per-run rating deltas ---
    hist = runners_df.copy()
    hist['date'] = pd.to_datetime(hist['date'], errors='coerce')
    hist = hist.sort_values(['horseId', 'date'])

    # Per horse, change in rating_after_race compared to previous start
    hist['run_to_rating'] = (hist
                            .groupby('horseId', group_keys=False)['rating_after_race']
                            .apply(lambda s: s.diff()))

    # Historical data within 365-day window
    historical_data_365 = hist[
        (hist['date'] >= cutoff_date_365) &
        (hist['date'] < race_date)
    ].copy()

    # All historical data (for going/distance metrics)
    historical_data_all = hist[hist['date'] < race_date].copy()

    # --- Calculate all metrics for each horse ---
    horse_metrics = []

    for _, horse_row in per_horse_today.iterrows():
        horse_id = horse_row['horseId']
        horse_name = horse_row['horseName']
        today_weight_kg = horse_row['weightKg']  # Get today's weight for adjusted rating calculation

        # Get all historical data for this horse (before race date only)
        horse_hist_365 = historical_data_365[
            historical_data_365['horseId'] == horse_id
        ].copy()

        horse_hist_all = historical_data_all[
            historical_data_all['horseId'] == horse_id
        ].copy()

        # --- Calculate runs/wins/places (365 days) ---
        if horse_hist_365.empty:
            runs_365 = 0
            wins_365 = 0
            places_365 = 0
            rtr = 0.0
        else:
            runs_365 = len(horse_hist_365)
            wins_365 = horse_hist_365['win'].sum()  # Sum of 1s in win column
            places_365 = horse_hist_365['place'].sum()  # Sum of 1s in place column

            # rtr mean (365 days, this horse only)
            rtr = horse_hist_365['run_to_rating'].dropna().mean()
            if pd.isna(rtr):
                rtr = 0.0

        # Format runs/wins/places for 365 days
        runs_wins_places_365d = f"{runs_365}/{wins_365}/{places_365}"

        # --- Calculate runs/wins/places (all time) ---
        if horse_hist_all.empty:
            runs_all = 0
            wins_all = 0
            places_all = 0
        else:
            runs_all = len(horse_hist_all)
            wins_all = horse_hist_all['win'].sum()  # Sum of 1s in win column
            places_all = horse_hist_all['place'].sum()  # Sum of 1s in place column

        # Format runs/wins/places for all time
        runs_wins_places = f"{runs_all}/{wins_all}/{places_all}"

        # Format rtr with runs count
        rtr_formatted = str(np.round(rtr, 2)) + ' (' + str(runs_365) + ')'

        # --- Calculate last adjusted rating and days since last run ---
        if not horse_hist_all.empty:
            # Get the most recent run before the race date
            last_run = horse_hist_all.iloc[-1]  # Most recent run (data is sorted by date)

            # CORRECTED: Use today's weight instead of last run's weight
            if pd.notna(last_run['rating_after_race']) and pd.notna(today_weight_kg):
                last_adjusted_rating = last_run['rating_after_race'] + 55 - today_weight_kg
            else:
                last_adjusted_rating = None

            # Calculate days since last run
            if pd.notna(last_run['date']):
                days_since_last_run = (race_date - last_run['date']).days
            else:
                days_since_last_run = None
        else:
            last_adjusted_rating = None
            days_since_last_run = None

        # --- Calculate rtr for last 5 runs ---
        if not horse_hist_all.empty:
            # Get the last 5 runs (or fewer if less than 5 available)
            last_5_runs = horse_hist_all.tail(5)
            last_5_ratings = last_5_runs['run_to_rating'].dropna()

            if len(last_5_ratings) > 0:
                # Format as individual values, most recent first (no sample size brackets)
                last_5_values = [str(np.round(val, 2)) for val in reversed(last_5_ratings.tolist())]
                rtr_last_5 = ', '.join(last_5_values)
            else:
                rtr_last_5 = '-'
        else:
            rtr_last_5 = '-'

        # --- Calculate additional rtr metrics ---

        # 1. Same going category for this horse (all time)
        if race_going_category is not None:
            going_data = horse_hist_all[
                horse_hist_all['going_category'] == race_going_category
            ]
            going_runs = len(going_data['run_to_rating'].dropna()) if not going_data.empty else 0
            rtr_going_category = going_data['run_to_rating'].dropna().mean() if not going_data.empty else 0.0
        else:
            going_runs = 0
            rtr_going_category = 0.0

        # 2. Same distance group for this horse (all time)
        if race_distance_group is not None:
            distance_data = horse_hist_all[
                horse_hist_all['distance_group'] == race_distance_group
            ]
            distance_runs = len(distance_data['run_to_rating'].dropna()) if not distance_data.empty else 0
            rtr_distance_group = distance_data['run_to_rating'].dropna().mean() if not distance_data.empty else 0.0
        else:
            distance_runs = 0
            rtr_distance_group = 0.0

        horse_metrics.append({
            'horseId': horse_id,
            'horseName': horse_name,
            'runs_wins_places_365d': runs_wins_places_365d,
            'runs_wins_places': runs_wins_places,
            'rtr': rtr_formatted,
            'rtr_going_category': rtr_going_category,
            'rtr_distance_group': rtr_distance_group,
            'rtr_last_5': rtr_last_5,
            'adjusted_rar': last_adjusted_rating,
            'days_since_last_run': days_since_last_run,
            'going_runs': going_runs,
            'distance_runs': distance_runs
        })

    horse_metrics_df = pd.DataFrame(horse_metrics)

    # Format the going and distance metrics with sample sizes in brackets
    horse_metrics_df['rtr_going_category'] = (
        np.round(horse_metrics_df['rtr_going_category'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['going_runs'].astype(str) + ')'
    )

    horse_metrics_df['rtr_distance_group'] = (
        np.round(horse_metrics_df['rtr_distance_group'].fillna(0.0), 2).astype(str) +
        ' (' + horse_metrics_df['distance_runs'].astype(str) + ')'
    )

    # --- Prepare final output ---
    out = (per_horse_today
           .merge(horse_metrics_df[['horseId', 'runs_wins_places_365d', 'runs_wins_places', 'rtr',
                                   'rtr_going_category', 'rtr_distance_group',
                                   'rtr_last_5', 'adjusted_rar', 'days_since_last_run']],
                  on='horseId', how='left')
           .fillna({
               'runs_wins_places_365d': '0/0/0', 'runs_wins_places': '0/0/0', 'rtr': '0.0 (0)',
               'rtr_going_category': '0.0 (0)', 'rtr_distance_group': '0.0 (0)',
               'rtr_last_5': '-'
           }))

    # --- Order & select final columns ---
    out = out[['horseId', 'horseName', 'sex', 'age', 'tongueTie', 'hood', 'blinkers', 'adjusted_handicapRatingKg', 'adjusted_rar',
               'runs_wins_places_365d', 'runs_wins_places', 'rtr', 'rtr_going_category',
               'rtr_distance_group', 'rtr_last_5', 'days_since_last_run']].sort_values(by='adjusted_rar', ascending=False)

    # Ensure proper data types
    out['rtr'] = out['rtr'].astype(str)
    out['runs_wins_places_365d'] = out['runs_wins_places_365d'].astype(str)
    out['runs_wins_places'] = out['runs_wins_places'].astype(str)
    out['rtr_going_category'] = out['rtr_going_category'].astype(str)
    out['rtr_distance_group'] = out['rtr_distance_group'].astype(str)
    out['rtr_last_5'] = out['rtr_last_5'].astype(str)

    return out
```

```python
import pandas as pd
import numpy as np

def calculate_standard_times(races, method='median', percentile=10, min_races=5):
    """
    Calculate standard times for horse racing data using industry best practices.

    Parameters:
    -----------
    races : pd.DataFrame
        DataFrame containing racing data with columns:
        - name_meeting: track/meeting name
        - distance: race distance
        - direction: race direction (e.g., 'L', 'R')
        - rail: rail position
        - surface: track surface type
        - winnerTime_seconds: winning time in seconds

    method : str, default='median'
        Method to calculate standard time:
        - 'median': Use median of winning times
        - 'mean': Use mean of winning times
        - 'percentile': Use specified percentile
        - 'trimmed_mean': Use trimmed mean (excluding top/bottom 10%)
        - 'weighted_recent': Weight recent races more heavily

    percentile : int, default=10
        Percentile to use if method='percentile' (10 = fastest 10%)

    min_races : int, default=5
        Minimum number of races required to calculate a standard time

    Returns:
    --------
    pd.DataFrame with standard times and statistics per combination
    """

    # Group by the specified combinations
    grouping_cols = ['name_meeting', 'distance', 'direction', 'rail', 'surface']

    # Ensure all grouping columns exist
    missing_cols = [col for col in grouping_cols if col not in races.columns]
    if missing_cols:
        print(f"Warning: Missing columns {missing_cols}. Adjusting grouping...")
        grouping_cols = [col for col in grouping_cols if col in races.columns]

    # Create a copy to avoid modifying original data
    df = races.copy()

    # Remove invalid times (nulls, zeros, or negative values)
    df = df[df['winnerTime_seconds'].notna()]
    df = df[df['winnerTime_seconds'] > 0]

    # Calculate statistics for each combination
    results = []

    for group_keys, group_data in df.groupby(grouping_cols):
        n_races = len(group_data)

        # Skip if insufficient data
        if n_races < min_races:
            continue

        times = group_data['winnerTime_seconds'].values

        # Calculate standard time based on method
        if method == 'median':
            standard_time = np.median(times)

        elif method == 'mean':
            standard_time = np.mean(times)

        elif method == 'percentile':
            # Lower percentile = faster times (better performances)
            standard_time = np.percentile(times, percentile)

        elif method == 'trimmed_mean':
            # Remove top and bottom 10% before calculating mean
            lower = np.percentile(times, 10)
            upper = np.percentile(times, 90)
            trimmed_times = times[(times >= lower) & (times <= upper)]
            standard_time = np.mean(trimmed_times) if len(trimmed_times) > 0 else np.mean(times)

        elif method == 'weighted_recent':
            # Weight more recent races more heavily (assuming data is sorted by date)
            weights = np.linspace(0.5, 1.0, n_races)
            weights = weights / weights.sum()
            standard_time = np.average(times, weights=weights)

        else:
            raise ValueError(f"Unknown method: {method}")

        # Calculate additional statistics
        result = {col: group_keys[i] for i, col in enumerate(grouping_cols)}
        result.update({
            'standard_time': round(standard_time, 2),
            'n_races': n_races,
            'min_time': round(np.min(times), 2),
            'max_time': round(np.max(times), 2),
            'mean_time': round(np.mean(times), 2),
            'median_time': round(np.median(times), 2),
            'std_dev': round(np.std(times), 2),
            'coef_variation': round(np.std(times) / np.mean(times) * 100, 2),  # CV in %
            'p10_time': round(np.percentile(times, 10), 2),  # Fastest 10%
            'p25_time': round(np.percentile(times, 25), 2),  # First quartile
            'p75_time': round(np.percentile(times, 75), 2),  # Third quartile
            'p90_time': round(np.percentile(times, 90), 2),  # Slowest 10%
        })

        results.append(result)

    # Create results DataFrame
    standard_times_df = pd.DataFrame(results)

    if len(standard_times_df) == 0:
        print(f"Warning: No combinations had at least {min_races} races")
        return pd.DataFrame()

    # Sort by grouping columns for better readability
    standard_times_df = standard_times_df.sort_values(grouping_cols)

    return standard_times_df


def calculate_going_corrections(races, standard_times_df, method='deviation',
                               min_races_per_day=3, outlier_threshold=2.5):
    """
    Calculate going corrections for each race day using industry best practices.

    Going correction represents how much faster/slower the track is running
    compared to standard times due to track conditions (firm, good, soft, heavy, etc.)

    Parameters:
    -----------
    races : pd.DataFrame
        DataFrame containing racing data with columns:
        - name_meeting: track/meeting name
        - distance: race distance
        - direction: race direction
        - rail: rail position
        - surface: track surface type
        - going: track condition
        - winnerTime_seconds: winning time in seconds
        - date: race date

    standard_times_df : pd.DataFrame
        DataFrame with pre-calculated standard times for each track/distance combination

    method : str, default='deviation'
        Method to calculate going correction:
        - 'deviation': Average deviation from standard times
        - 'weighted_deviation': Weight by class/quality of race
        - 'median_deviation': Use median to reduce outlier impact
        - 'regression': Use regression analysis for correction

    min_races_per_day : int, default=3
        Minimum races required on a day to calculate reliable correction

    outlier_threshold : float, default=2.5
        Standard deviations from mean to consider as outlier

    Returns:
    --------
    DataFrame with going corrections per race and aggregated by day/meeting
    """

    # Ensure we have necessary columns
    required_cols = ['name_meeting', 'distance', 'direction', 'rail', 'surface',
                     'going', 'winnerTime_seconds']
    missing_cols = [col for col in required_cols if col not in races.columns]
    if missing_cols:
        # Adjust for missing columns
        if 'direction' not in races.columns:
            races['direction'] = 'N/A'
        if 'rail' not in races.columns:
            races['rail'] = 'N/A'

    # Add date column if not present (extract from datetime if needed)
    if 'date' not in races.columns:
        if 'datetime' in races.columns:
            races['date'] = pd.to_datetime(races['datetime']).dt.date
        else:
            races['date'] = pd.Timestamp.now().date()

    # Merge with standard times
    merge_cols = ['name_meeting', 'distance', 'direction', 'rail', 'surface']
    merge_cols = [col for col in merge_cols if col in races.columns and col in standard_times_df.columns]

    df = races.merge(
        standard_times_df[merge_cols + ['standard_time']],
        on=merge_cols,
        how='left'
    )

    # Calculate raw deviation from standard time for each race
    df['raw_deviation'] = df['winnerTime_seconds'] - df['standard_time']
    df['deviation_percent'] = (df['raw_deviation'] / df['standard_time']) * 100

    # Remove races without standard times
    df = df[df['standard_time'].notna()]

    # Group by meeting, date, and going to calculate corrections
    grouping = ['name_meeting', 'date', 'going']

    results = []

    for group_keys, group_data in df.groupby(grouping):
        n_races = len(group_data)

        if n_races < min_races_per_day:
            continue

        deviations = group_data['raw_deviation'].values
        deviations_pct = group_data['deviation_percent'].values

        # Remove outliers using IQR method (more robust than z-score for small samples)
        Q1 = np.percentile(deviations, 25)
        Q3 = np.percentile(deviations, 75)
        IQR = Q3 - Q1
        lower_bound = Q1 - outlier_threshold * IQR
        upper_bound = Q3 + outlier_threshold * IQR

        # Filter outliers
        mask = (deviations >= lower_bound) & (deviations <= upper_bound)
        filtered_deviations = deviations[mask]
        filtered_deviations_pct = deviations_pct[mask]

        if len(filtered_deviations) < min_races_per_day:
            # If too many outliers removed, use original with less strict filtering
            mask = (deviations >= Q1 - 3*IQR) & (deviations <= Q3 + 3*IQR)
            filtered_deviations = deviations[mask]
            filtered_deviations_pct = deviations_pct[mask]

        # Calculate going correction based on method
        if method == 'deviation':
            going_correction = np.mean(filtered_deviations)
            going_correction_pct = np.mean(filtered_deviations_pct)

        elif method == 'median_deviation':
            going_correction = np.median(filtered_deviations)
            going_correction_pct = np.median(filtered_deviations_pct)

        elif method == 'weighted_deviation':
            # Weight by inverse of standard deviation (more consistent times get more weight)
            weights = 1 / (group_data['std_dev'].values[mask] + 0.1)  # Add small constant to avoid division by zero
            weights = weights / weights.sum()
            going_correction = np.average(filtered_deviations, weights=weights)
            going_correction_pct = np.average(filtered_deviations_pct, weights=weights)

        else:  # Default to mean
            going_correction = np.mean(filtered_deviations)
            going_correction_pct = np.mean(filtered_deviations_pct)

        result = {
            'name_meeting': group_keys[0],
            'date': group_keys[1],
            'going': group_keys[2],
            'going_correction_seconds': round(going_correction, 2),
            'going_correction_percent': round(going_correction_pct, 2),
            'n_races': n_races,
            'n_races_used': len(filtered_deviations),
            'n_outliers_removed': n_races - len(filtered_deviations),
            'min_deviation': round(np.min(deviations), 2),
            'max_deviation': round(np.max(deviations), 2),
            'std_deviation': round(np.std(filtered_deviations), 2),
            'confidence': calculate_confidence_score(
                n_races=len(filtered_deviations),
                std_dev=np.std(filtered_deviations),
                outliers_removed=n_races - len(filtered_deviations)
            )
        }

        results.append(result)

    going_corrections_df = pd.DataFrame(results)

    if len(going_corrections_df) == 0:
        print(f"Warning: No day/meeting combinations had at least {min_races_per_day} races")
        return pd.DataFrame()

    # Add going category interpretation
    going_corrections_df['track_speed'] = going_corrections_df['going_correction_seconds'].apply(
        lambda x: 'Fast' if x < -0.5 else 'Slow' if x > 0.5 else 'Standard'
    )

    # Sort by date and meeting
    going_corrections_df = going_corrections_df.sort_values(['date', 'name_meeting'])

    return going_corrections_df


def calculate_confidence_score(n_races, std_dev, outliers_removed):
    """
    Calculate a confidence score (0-100) for the going correction.

    Higher scores indicate more reliable corrections.
    """
    # Base score from number of races (max 40 points)
    races_score = min(40, n_races * 4)

    # Consistency score from standard deviation (max 40 points)
    # Lower std_dev is better
    if std_dev < 0.5:
        consistency_score = 40
    elif std_dev < 1.0:
        consistency_score = 30
    elif std_dev < 2.0:
        consistency_score = 20
    else:
        consistency_score = 10

    # Outlier penalty (max 20 points)
    outlier_ratio = outliers_removed / (n_races + outliers_removed) if n_races > 0 else 0
    outlier_score = 20 * (1 - outlier_ratio)

    total_score = races_score + consistency_score + outlier_score

    return round(min(100, total_score))


def apply_going_corrections_to_races(races, going_corrections_df):
    """
    Apply going corrections to individual race times to get corrected times.

    Parameters:
    -----------
    races : pd.DataFrame
        Original races data
    going_corrections_df : pd.DataFrame
        Calculated going corrections

    Returns:
    --------
    DataFrame with original times and corrected times
    """
    # Ensure date column exists
    if 'date' not in races.columns:
        if 'datetime' in races.columns:
            races['date'] = pd.to_datetime(races['datetime']).dt.date
        else:
            races['date'] = pd.Timestamp.now().date()

    # Merge corrections with races
    df = races.merge(
        going_corrections_df[['name_meeting', 'date', 'going', 'going_correction_seconds',
                              'going_correction_percent', 'track_speed', 'confidence']],
        on=['name_meeting', 'date', 'going'],
        how='left'
    )

    # Apply correction (subtract because positive correction means track is slow)
    df['corrected_time'] = df['winnerTime_seconds'] - df['going_correction_seconds']

    # Round for readability
    df['corrected_time'] = df['corrected_time'].round(2)

    # Add flag for whether correction was applied
    df['correction_applied'] = df['going_correction_seconds'].notna()

    return df


def analyze_going_impact(going_corrections_df, races=None):
    """
    Analyze the impact of different going conditions, optionally by distance.

    Parameters:
    -----------
    going_corrections_df : pd.DataFrame
        Calculated going corrections
    races : pd.DataFrame, optional
        Original races data to include distance analysis

    Returns:
    --------
    Summary statistics by going condition (and distance if races provided)
    """
    # Basic summary by going
    summary = going_corrections_df.groupby('going').agg({
        'going_correction_seconds': ['mean', 'median', 'std', 'min', 'max'],
        'going_correction_percent': ['mean', 'median'],
        'n_races': 'sum',
        'confidence': 'mean'
    }).round(2)

    summary.columns = ['_'.join(col).strip() for col in summary.columns.values]

    # Add interpretation
    summary['typical_impact'] = summary['going_correction_seconds_mean'].apply(
        lambda x: f"{'Faster' if x < 0 else 'Slower'} by {abs(x):.1f}s"
    )

    return summary


def analyze_going_impact_by_distance(races, going_corrections_df):
    """
    Analyze going impact broken down by distance categories.

    Parameters:
    -----------
    races : pd.DataFrame
        Original races data with distance information
    going_corrections_df : pd.DataFrame
        Calculated going corrections

    Returns:
    --------
    Summary statistics by going condition and distance category
    """
    # Ensure date column exists for merging
    if 'date' not in races.columns:
        if 'datetime' in races.columns:
            races['date'] = pd.to_datetime(races['datetime']).dt.date

    # Merge going corrections with races to get distance info
    df = races.merge(
        going_corrections_df[['name_meeting', 'date', 'going',
                              'going_correction_seconds', 'going_correction_percent']],
        on=['name_meeting', 'date', 'going'],
        how='inner'
    ).drop_duplicates(subset=['name_meeting', 'date', 'going', 'distance'])

    # Create distance categories
    df['distance_category'] = pd.cut(
        df['distance'],
        bins=[0, 1200, 1600, 2000, 2400, 9999],
        labels=['Sprint (≤1200m)', 'Mile (1201-1600m)', 'Middle (1601-2000m)',
                'Long (2001-2400m)', 'Extended (>2400m)']
    )

    # Calculate summary by going and distance
    summary = df.groupby(['going', 'distance_category']).agg({
        'going_correction_seconds': ['mean', 'median', 'std'],
        'going_correction_percent': ['mean', 'median'],
        'distance': ['count', 'mean']
    }).round(2)

    summary.columns = ['_'.join(col).strip() for col in summary.columns.values]
    summary = summary.rename(columns={'distance_count': 'n_observations',
                                     'distance_mean': 'avg_distance'})

    # Calculate the scaling factor (how correction increases with distance)
    pivot_seconds = summary['going_correction_seconds_mean'].unstack(fill_value=0)
    pivot_percent = summary['going_correction_percent_mean'].unstack(fill_value=0)

    return summary, pivot_seconds, pivot_percent


def calculate_distance_adjusted_going_corrections(races, standard_times_df,
                                                 base_distance=1400,
                                                 method='scaled_deviation',
                                                 min_races_per_day=3):
    """
    Calculate going corrections that properly account for distance effects.

    This method normalizes corrections to a base distance, then scales appropriately.

    Parameters:
    -----------
    races : pd.DataFrame
        Racing data
    standard_times_df : pd.DataFrame
        Pre-calculated standard times
    base_distance : int, default=1400
        Reference distance for normalization (meters)
    method : str, default='scaled_deviation'
        Calculation method
    min_races_per_day : int, default=3
        Minimum races required

    Returns:
    --------
    DataFrame with distance-adjusted going corrections
    """
    # First get standard going corrections
    base_corrections = calculate_going_corrections(
        races, standard_times_df, method='median_deviation',
        min_races_per_day=min_races_per_day
    )

    if len(base_corrections) == 0:
        return pd.DataFrame()

    # Merge with races to get distances
    if 'date' not in races.columns:
        if 'datetime' in races.columns:
            races['date'] = pd.to_datetime(races['datetime']).dt.date

    corrections_with_distance = races[['name_meeting', 'date', 'going', 'distance']].drop_duplicates().merge(
        base_corrections,
        on=['name_meeting', 'date', 'going'],
        how='inner'
    )

    # Calculate the distance scaling factor
    # Going impact typically scales roughly linearly with distance
    corrections_with_distance['distance_factor'] = corrections_with_distance['distance'] / base_distance

    # Adjust the going correction to base distance
    corrections_with_distance['normalized_correction_seconds'] = (
        corrections_with_distance['going_correction_seconds'] /
        corrections_with_distance['distance_factor']
    )

    # Group by going to get the average normalized correction
    going_factors = corrections_with_distance.groupby('going').agg({
        'normalized_correction_seconds': 'median',
        'going_correction_percent': 'median',  # Percentage should be more consistent
        'n_races': 'sum'
    }).round(3)

    # Create a lookup table for different distances
    distances = [1000, 1200, 1400, 1600, 1800, 2000, 2400, 3000]

    going_by_distance = []
    for going in going_factors.index:
        base_correction = going_factors.loc[going, 'normalized_correction_seconds']
        pct_correction = going_factors.loc[going, 'going_correction_percent']

        for dist in distances:
            going_by_distance.append({
                'going': going,
                'distance': dist,
                'expected_correction_seconds': round(base_correction * (dist / base_distance), 2),
                'expected_correction_percent': round(pct_correction, 2)
            })

    going_distance_df = pd.DataFrame(going_by_distance)

    return corrections_with_distance, going_distance_df


```
